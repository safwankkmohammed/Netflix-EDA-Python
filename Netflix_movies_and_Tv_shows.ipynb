## Handling Missing Data

Visualizing Missing data

```python
plt.title("count of missing data")
missing_values = df.isnull().sum().sort_values(ascending=False)
plt.bar(missing_values.index, missing_values.values)
plt.xlabel("columns")
plt.ylabel("Mssing values")
plt.xticks(rotation=90)
plt.show()
```

```python
df["director"] = df["director"].fillna("Not Available")
```

```python
df['cast'] = df['cast'].fillna('Not Available')
```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
df['rating'] = df['rating'].fillna('Not Rated')
```

```python
df['duration'] = df['duration'].fillna('Not Available')
```

```python
df.dropna(subset=['date_added'], inplace=True)
```

```python
df.isnull().sum()
```

## Data cleaning & Transformation

converting "date_added" column to datetime format


```python
# Strip spaces before conversion
df['date_added'] = df['date_added'].str.strip()

# Convert to datetime safely
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

```

```python
df['date_added'].head()
df['date_added'].isna().sum()
```

Fixing Incorrect Rating Values

```python
df['rating'].value_counts()
```

```python
# Removing incorrect rating values that contain 'min'
df = df[~df['rating'].str.contains('min', case=False, na=False)]

```

```python
df['rating'].value_counts()
```

## Feature Engineering

```python
# Extract year and month
df.loc[:, 'year_added'] = df['date_added'].dt.year
df.loc[:, 'month_added'] = df['date_added'].dt.month

```

```python
df
```

```python
# Convert to lowercase and create a helper column
df.loc[:, 'type'] = df['type'].str.lower()
df.loc[:, 'content_category'] = df['type'].apply(lambda x: 'Movie' if x == 'movie' else 'TV Show')
```

```python
text_cols = ['country', 'rating', 'listed_in']
for col in text_cols:
    df[col] = df[col].str.strip().str.lower()
```

```python
df[['duration_int', 'duration_unit']] = df['duration'].str.extract('(\d+)\s*(\w+)')
df['duration_int'] = pd.to_numeric(df['duration_int'], errors='coerce')
```

## Exploratory Data Analysis

Type Distribution — Movies vs TV Shows

```python
type_counts = df['type'].value_counts()

plt.figure(figsize=(6,5))
sns.barplot(x=type_counts.index, y=type_counts.values, palette='Set2')
plt.title("Distribution of Content Type")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()

print(type_counts)
```

Release Year Distribution

```python
plt.figure(figsize=(10,5))
sns.histplot(df['release_year'], bins=30, kde=True, color='teal')
plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

Top 10 Countries with Most Content

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(y=top_countries.index, x=top_countries.values, palette='mako')
plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()

print(top_countries)
```

Top 10 Genres

```python
from collections import Counter

# Split multiple genres
genres = df['listed_in'].str.split(', ')
all_genres = [genre for sublist in genres for genre in sublist]
genre_counts = Counter(all_genres).most_common(10)

# Convert to DataFrame for plotting
genre_df = pd.DataFrame(genre_counts, columns=['Genre', 'Count'])

plt.figure(figsize=(10,6))
sns.barplot(y='Genre', x='Count', data=genre_df, palette='viridis')
plt.title("Top 10 Genres on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()

genre_df
```
